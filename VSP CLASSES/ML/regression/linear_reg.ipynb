```python
## Implementataion of simple linear regression

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
#sklearn--scikit-learn
```

```python
mydf = pd.read_csv("robot_battery_data.csv")

mydf.info()
```

```python
mydf.head()
```

```python
mydf.tail()
```

```python
mydf.shape
```

```python
X_axis = mydf[['payload_kg']]
```

```python
X_axis.head()
```

```python
y_axis = mydf[['battery_drain']]
```

```python
y_axis.head()
```

```python
my_model = LinearRegression().fit(X_axis, y_axis)
# fit is used for training the model, in our case model is LinearRegression
```

```python
my_model
```

```python
print(my_model.coef_)
```

```python
print(my_model.intercept_)
```

```python
my_model.predict([[10]])
```

```python
#my_model is the final model
```

```python
sns.lineplot(data = mydf, x = 'payload_kg', y = 'battery_drain')
plt.show()
```
